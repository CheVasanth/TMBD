```python
#Finding the IQR
percentile25 = df1['budget'].quantile(0.25)
percentile75 = df1['budget'].quantile(0.75)

iqr=percentile75 - percentile25
print(iqr)
print(percentile25)
print(percentile75)
```

```python
#Finding upper and lower limit
upper_limit = percentile75 + 1.5 * iqr
lower_limit = percentile25 - 1.5 * iqr
print(upper_limit)
print(lower_limit)
```
